```python
from opt.opt import opt_unconstrained
```

```python
def rosen(x):
    return 100.0 * (x[1] - x[0]**2)**2 + (1.0 - x[0])**2

def rosen_grad(x):
    return np.array([-400.0 * x[0] * (x[1] - x[0]**2) - 2.0 * (1.0 - x[0]),
                      200.0 * (x[1] - x[0]**2)])

def rosen_hess(x):
    return np.array([[1200.0 * x[0]**2 - 400.0 * x[1] + 2.0, -400.0 * x[0]],
                     [-400.0 * x[0],                          200.0]])        
```

```python
prob = opt_unconstrained(rosen, rosen_grad, x0=[-1.2, 1.0], nombre="Rosenbrock")
print("f(x0)    =", prob.f(prob.x0))
print("grad(x*) =", rosen_grad(np.array([1.0, 1.0])))
```

```python
from opt.opt import verificar_gradiente
rng = np.random.default_rng(0)
for _ in range(3):
    x = rng.normal(size=2)
    print(f"x = {x},  error relativo = {verificar_gradiente(rosen, rosen_grad, x):.2e}")
```
